# Fase 4 — Controlar el flujo: break, continue, for...else

# 4.1 break — salir cuando aciertas



```python
for i in range(1, 11):
    print(i)
    if i == 5:
        print("Encontré el 5, me detengo.")
        break
# Imprime 1, 2, 3, 4, 5 y luego se detiene.
```

# 4.2 continue — saltar esta vuelta



```python
for n in [3, -1, 7, -4, 2]:
    if n < 0:
        continue   # ignora los negativos
    print(n)
# Imprime: 3, 7, 2
```

# 4.3 for...else — "se acabaron los intentos"



```python
numbers = [2, 4, 6, 8, 10]
for n in numbers:
    if n == 7:
        print("Encontré el 7")
        break
else:
    print("El 7 no estaba en la lista.")
# Output: El 7 no estaba en la lista.
```

# 4.4 Versión 2 del juego — con control de flujo



```python
secret_number = 42
max_attempts = 7

for attempt_num in range(1, max_attempts + 1):
    entry = input(f"Intento {attempt_num}/{max_attempts} — Tu número (1-100): ")

    # Validación 1: ¿es número?
    if not entry.isnumeric():
        print("  Eso no es un número. Esta vuelta no cuenta.")
        continue

    guess = int(entry)

    # Validación 2: ¿está en rango?
    if not (1 <= guess <= 100):
        print("  El número debe estar entre 1 y 100. Esta vuelta no cuenta.")
        continue

    # Lógica del juego
    if guess == secret_number:
        print(f"¡Correcto! Adivinaste en {attempt_num} intento(s).")
        break
    elif guess > secret_number:
        print("  Muy alto.")
    else:
        print("  Muy bajo.")
else:
    print(f"Se acabaron los intentos. El número era {secret_number}.")
```

# 5.3 Versión final del juego



```python
# Juego de Adivina el Número — Python Mini-Bootcamp Semana 2
import random

print("=" * 40)
print("    ADIVINA EL NÚMERO")
print("=" * 40)

play_again = True

while play_again:
    # --- Configuración de la partida ---
    secret_number = random.randint(1, 100)
    max_attempts = 7
    valid_attempts = 0
    guessed_right = False

    print(f"\nEstoy pensando en un número entre 1 y 100.")
    print(f"Tienes {max_attempts} intentos. ¡Buena suerte!\n")

    # --- Ciclo principal del juego ---
    while valid_attempts < max_attempts:
        remaining = max_attempts - valid_attempts
        entry = input(f"Intento {valid_attempts + 1}/{max_attempts} — Tu número: ").strip()

        # Validación 1: ¿es número?
        if not entry.isnumeric():
            print("  Eso no es un número. Esta vuelta no cuenta.\n")
            continue

        guess = int(entry)

        # Validación 2: ¿está en rango?
        if not (1 <= guess <= 100):
            print("  El número debe estar entre 1 y 100. Esta vuelta no cuenta.\n")
            continue

        # Si llegó aquí, el intento es válido
        valid_attempts += 1

        # --- Lógica del juego ---
        if guess == secret_number:
            print(f"\n¡Correcto! Adivinaste en {valid_attempts} intento(s).")
            guessed_right = True
            break
        elif guess > secret_number:
            print(f"  Muy alto. Te quedan {remaining - 1 if remaining > 0 else 0} intento(s).\n")
        else:
            print(f"  Muy bajo. Te quedan {remaining - 1 if remaining > 0 else 0} intento(s).\n")

    # --- Resultado de la partida ---
    if not guessed_right:
        print(f"\nSe acabaron los intentos. El número era {secret_number}.")

    # --- Jugar de nuevo ---
    print()
    answer = input("¿Quieres jugar de nuevo? (si/no): ").strip().lower()
    play_again = (answer == "si")

print("\n¡Gracias por jugar! Hasta la próxima.")
```

# Reto B — Pista de "frío/caliente"

```python
# Juego de Adivina el Número — Python Mini-Bootcamp Semana 2
import random
import sys

print("=" * 40)
print("    ADIVINA EL NÚMERO")
print("=" * 40)
print()
print("Facil, Normal o Dificil:")

difficult = input().lower()
if difficult == "facil" or difficult[0].lower() == "f":
  max_attempts = 10
  random_max = 50
elif difficult == "normal" or difficult[0].lower() == "n":
  max_attempts = 7
  random_max = 100
else: # Dificil
  max_attempts = 6
  random_max = 200

play_again = True

while play_again:
    # --- Configuración de la partida ---
    secret_number = random.randint(1, max_attempts)
    valid_attempts = 0
    guessed_right = False

    print(f"\nEstoy pensando en un número entre 1 y {random_max}.")
    print(f"Tienes {max_attempts} intentos. ¡Buena suerte!\n")

    # --- Ciclo principal del juego ---
    while valid_attempts < max_attempts:
        remaining = max_attempts - valid_attempts
        entry = input(f"Intento {valid_attempts + 1}/{max_attempts} — Tu número: ").strip()

        # Validación 1: ¿es número?
        if not entry.isnumeric():
            print("  Eso no es un número. Esta vuelta no cuenta.\n")
            continue

        guess = int(entry)

        # Validación 2: ¿está en rango?
        if not (1 <= guess <= random_max):
            print(f"  El número debe estar entre 1 y {random_max}. Esta vuelta no cuenta.\n")
            continue

        # Si llegó aquí, el intento es válido
        valid_attempts += 1

        # --- Lógica del juego ---

        if guess == secret_number:
            print(f"\n¡Correcto! Adivinaste en {valid_attempts} intento(s).")
            guessed_right = True
            break
        else:
            difference = abs(guess - secret_number)
            if difference <= 3:
               hint = "¡Caliente!"
            elif difference <= 10:
               hint = "Tibio"
            else:
               hint = "Frío"

            if guess > secret_number:
               print(f"  Muy alto. {hint}. Te quedan {remaining - 1 if remaining > 0 else 0} intento(s).\n")
            else:
               print(f"  Muy bajo. {hint}. Te quedan {remaining - 1 if remaining > 0 else 0} intento(s).\n")

    # --- Resultado de la partida ---
    if not guessed_right:
        print(f"\nSe acabaron los intentos. El número era {secret_number}.")

    # --- Jugar de nuevo ---
    print()
    answer = input("¿Quieres jugar de nuevo? (si/no): ").strip().lower()
    play_again = (answer == "si")

print("\n¡Gracias por jugar! Hasta la próxima.")
```
